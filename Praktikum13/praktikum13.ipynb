Import Library

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import RobustScaler1
from sklearn.model_selection import train_test_split
```

Load Dataset

```python
df = pd.read_csv('/content/drive/MyDrive/praktikum_ml/praktikum13/data/titanic_data.csv')
df.head()
```

Pemeriksaan Missing Values (Data Kosong).

```python
df.isnull().sum()
```

Data Cleaning

```python
#drop kolom
df = df.drop(columns=['Cabin', 'Name', 'Ticket'], axis=1)
```

Visualisasi Distribusi dan Outlier pada Kolom Age

```python
df.boxplot(column=['Age'])
```

 Mengisi Nilai Kosong pada Kolom Age dengan Median

```python
df['Age'].fillna(df['Age'].median(), inplace=True)
```

Mengisi Nilai Kosong pada Kolom Embarked dengan Modus

```python
df['Embarked'].fillna(df['Embarked'].mode()[0], inplace= True)
```

```python
df.isnull().sum()
```

Encoding Data Kategorikal

```python
replacements = {
    'Sex': {'male': 0, 'female': 1},
    'Embarked': {'S': 0, 'C': 1, 'Q': 2}
}

df.replace(replacements, inplace=True)
```

```python
df.head()
```

 Split Data dan Transformasi (Scaling)

```python
X = df.drop('Survived', axis=1)
y = df['Survived']
```

Transformasi Data (Scaling)

```python
from sklearn.preprocessing import RobustScaler # Perlu diimpor jika belum
scaler = RobustScaler()
X_scaled = scaler.fit_transform(X)
```

 Membagi Data Latih dan Data Uji

```python
from sklearn.model_selection import train_test_split # Perlu diimpor jika belum
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)
```

Modelling

```python
X_train.shape
```

```python
X_train.shape[1]
```

Pembuatan Model Neural Network

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.layers import Dropout

model = Sequential()

# Hidden layer 1
model.add(Dense(32, activation='relu', input_shape=(X_train.shape[1],)))

# Dropout untuk mengurangi overfitting
model.add(Dropout(0.3))

# Hidden layer 2
model.add(Dense(16, activation='relu'))
model.add(Dropout(0.3))

# Output layer (binary classification)
model.add(Dense(1, activation='sigmoid'))
```

 Kompilasi Model dan Ringkasan Arsitektur (Model Summary).

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

Early Stopping (Pencegahan Overfitting Selama Training)

```python
from tensorflow.keras.callbacks import EarlyStopping

early_stop = EarlyStopping(
    monitor='val_loss',
    patience=5,
    restore_best_weights=True
)
```

 Training Model Neural Network

```python
history = model.fit(
    X_train,
    y_train,
    epochs=100,
    batch_size=16,
    validation_split=0.2,
    callbacks=[early_stop],
    verbose=1
)
```

Evaluasi Model

```python
loss, accuracy = model.evaluate(X_test, y_test)

print("Test Loss :", loss)
print("Test Accuracy :", accuracy)
```

Visualisasi Hasil Pelatihan

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,4))

# Accuracy
plt.subplot(1,2,1)
plt.plot(history.history['accuracy'], label='Train Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.title('Model Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()

# Loss
plt.subplot(1,2,2)
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Model Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()

plt.show()
```
