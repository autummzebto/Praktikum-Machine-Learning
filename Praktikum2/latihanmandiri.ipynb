```python
import pandas as pd

# Baca dataset
df = pd.read_csv('/content/drive/MyDrive/praktikum_ml/tugas2/data/day.csv')
print("Total data :", len(df))

# (a) Data Training: 80% dari total dataset
train_size = int(0.8 * len(df))
train = df.iloc[:train_size]

# (c) Data Testing: 20% dari total dataset
test = df.iloc[train_size:]

# (b) Data Validation: 10% dari data training
val_size = int(0.1 * len(train))
final_train = train.iloc[:-val_size]
val = train.iloc[-val_size:]

# Tampilkan jumlah data
print("\nJumlah Data:")
print("Training   :", len(final_train))
print("Validation :", len(val))
print("Testing    :", len(test))

# Tampilkan 5 baris pertama
print("\nTraining (5 baris teratas):")
print(final_train.head())

print("\nValidation (5 baris teratas):")
print(val.head())

print("\nTesting (5 baris teratas):")
print(test.head())
```
